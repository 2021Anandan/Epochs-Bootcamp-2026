```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("dataset.csv")
df.head()
```

```python
print("Shape of dataset:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

df.info()
```

```python
df.isnull().sum()
```

```python
df.hist(figsize=(18,15))
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm", annot=False)
plt.title("Correlation Heatmap")
plt.show()
```

```python
df.columns.tolist()
```

```python
top_artists = df["artists"].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_artists.values, y=top_artists.index)

plt.title("Top 10 Artists")
plt.xlabel("Number of Songs")
plt.ylabel("Artist")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df["popularity"], bins=30, kde=True)

plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="danceability",
    y="popularity",
    alpha=0.5
)

plt.title("Danceability vs Popularity")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="energy",
    y="loudness",
    alpha=0.5
)

plt.title("Energy vs Loudness")
plt.show()
```

```python
top_genres = df["track_genre"].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title("Top 10 Genres")
plt.xlabel("Number of Tracks")
plt.ylabel("Genre")
plt.show()
```

```python
print("""
Key Insights

1. The dataset contains Spotify tracks with audio features and popularity scores.
2. Energy and loudness have a strong positive correlation.
3. Danceability has a moderate positive relationship with popularity.
4. Popularity is distributed unevenly, with many songs having medium popularity.
5. A few artists and genres contribute a large number of tracks.
6. Audio features can be useful for understanding song characteristics.
""")
```
